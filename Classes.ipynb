<span style="font-family:Arial; font-weight:Bold; font-size:1.8em; color:#00b3e5;"> Pet

```python
class Pet:
    
    n_Pets = 0
    
    @classmethod
    def add_n_Pets(cls): 
        cls.n_Pets = cls.n_Pets + 1
    
    @classmethod
    def get_n_Pets(cls): 
        return cls.n_Pets
        
        
    def __init__( self, name, age ):
        
        Pet.add_n_Pets()
        
        self.name = name
        self.age  = age        
        
        
    def get_name(self): return self.name
    def get_age(self) : return self.age
        
    def set_name(self,name): self.name = name
    def set_age( self,age ): self.age  = age
    
    
    def show(self):
        print(f"Hi ya! I am {self.get_name()}, {self.get_age()} years old. I am a pet!")
        

print("n_Pets =", Pet.get_n_Pets() )

p1 = Pet('Nemo',3)
p1.show()  
print("n_Pets =", Pet.get_n_Pets() )

p2 = Pet('Dory',9)
p2.show()
print("n_Pets =", Pet.get_n_Pets() )  
```

<span style="font-family:Arial; font-weight:Bold; font-size:1.8em; color:#00b3e5;"> Cat

```python
class Cat(Pet):
    
    n_Cats = 0
    
    @classmethod
    def add_n_Cats(cls):
        cls.n_Cats = cls.n_Cats+1
        
    @classmethod
    def get_n_Cats(cls):
        return cls.n_Cats
    
    
    def __init__( self, name, age, color ):
        
        Cat.add_n_Cats()
        
        super().__init__(name,age)
        self.color = color
        
        
    def get_color(self):
        return self.color
    
    def set_color(self,color):
        self.color = color
        
    def show(self):
        print(f"Hi ya! I am {self.get_name()}, {self.get_age()} years old. I am {self.get_color()}, I am a cat!")
        
        
        
c1 = Cat("GarField",5,"orange")
c1.show()
print("n_Cats =", Cat.get_n_Cats() )
print("n_Pets =", Pet.get_n_Pets() )
```
